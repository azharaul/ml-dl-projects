# 1. Instalasi dan Import

```python
from sentence_transformers import CrossEncoder

print("⏳ Memuat model Cross-Encoder...")
# Menggunakan model Cross-Encoder pre-trained untuk evaluasi relevansi
model = CrossEncoder("cross-encoder/ms-marco-MiniLM-L6-v2")
```

# 2. Reranking cross-encoder

```python
# 1. Query: Pertanyaan / pencarian gejala dari pengguna
query = "What are the early symptoms of hypertension and how is it diagnosed?"

# 2. Passages: Ringkasan artikel / literatur medis yang ada dalam database
passages = [
    "Hypertension, or high blood pressure, often has no early symptoms and is known as a silent killer. When symptoms occur, they may include headaches, shortness of breath, or nosebleeds.",
    "A definitive diagnosis of high blood pressure is made by measuring blood pressure using a sphygmomanometer over multiple doctor visits.",
    "Common symptoms of type 2 diabetes include increased thirst, frequent urination, unexplained weight loss, and fatigue.",
    "Diagnosis of skin lesions using convolutional neural networks involves evaluating dermatoscopic images for asymmetry, border irregularity, color, and diameter.",
    "Treatment for acute asthma attacks usually involves short-acting beta-agonists via an inhaler to relax airway muscles.",
    "To prevent hypertension, doctors recommend a low-sodium diet, regular cardiovascular exercise, maintaining a healthy weight, and limiting alcohol consumption."
]

print(f"\n🔍 Query Pasien/Pengguna: '{query}'\n")
print("=" * 70)

# 3. Melakukan Text Ranking / Reranking
ranks = model.rank(query, passages, return_documents=True)

# 4. Menampilkan hasil perangkingan
print("📊 Peringkat Artikel Medis Paling Relevan:\n")
for rank in ranks:
    print(f"Peringkat ID #{rank['corpus_id']} | Skor Relevansi: {rank['score']:.2f}")
    print(f"Ringkasan Teks: {rank['text']}\n")
```
